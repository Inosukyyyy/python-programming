Q1 Create a class Animal with:
method speak() → prints "Animal makes sound"
Create a child class Dog that:
inherits from Animal
call speak() using Dog object

```python
class animal:
    def speak(self):
        print("animal makes sound")

class dog(animal):
    pass

d = dog()
d.speak()
```

Q2 Create a class Vehicle:
method start() → prints "Vehicle started"
Create child class Car:
add method drive() → prints "Car is driving"

```python
class vehicle:
    def start(self):
        print("vehicle started")

class car(vehicle):
    def drive(self):
        print("car is driving")

c = car()
c .start()
c.drive()
```

Q3 Create a class Person:
constructor → name
Create child class Student:
inherits from Person
add variable marks
print both name and marks

```python
class person:
    def __init__(self, name):
        self.name = name

class student(person):
    def __init__(self,name, marks):
        super().__init__(name)
        self.marks = marks

s = student("Tanveer", 90)
print(s.name, s.marks)
```

Q4 Create a class Bird:
method fly() → prints "Bird can fly"
Create child class Penguin:
override fly() → prints "Penguin cannot fly"

```python
class bird:
    def fly(self):
        print("bird can fly")

class penguin(bird):
    def fly(self):
        print("penguin cannot")

p = penguin()
p.fly()
```
