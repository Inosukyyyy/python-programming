Q4 Create class Shape:
method area()
Create:
Rectangle → area = length × width
Circle → area = 3.14 × r × r

```python
class shape:
    def area(self):
        pass

class rectangle(shape):
    def __init__(self, l,w):
        self.l = l
        self.w = w

    def area(self):
        return self.l * self.w
    
class circle(shape):
    def __init__(self, r):
        self.r = r

    def area(self):
        return 3.14 * self.r * self.r

r = rectangle(10,5)
c = circle(7)

print(r.area())
print(c.area())
```
