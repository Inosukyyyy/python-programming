Q1 Create a class Product:
constructor → name, price
method apply_tax() → increase price by 18%
print final price

```python
class product:
    def __init__(self,name,price):
        self.name = name
        self.price = price

    def apply_tax(self):
        self.price += self.price * 0.18

p = product("shoes", 1000)
p.apply_tax()
print(p.price)
       
```
