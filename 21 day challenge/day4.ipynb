Q1 Create a class Calculator with:
method add(a, b)
method multiply(a, b)
call both methods

```python
class calculator:
    def add(self, a, b):
        print("sum", a+b)

    def multiply(self,a,b):
        print("multiply",a*b)

c = calculator()
c.add(5,3)
c.multiply(5,3)
```

Q2 Create a class Light with:
variable status = "OFF"
method turn_on()
method turn_off()
print status after each action

```python
class light:
    def __init__(self):
        self.status = "off"
    
    def turn_on(self):
        self.status = "on"
        print("light is", self.status)

    def turn_off(self):
        self.status = "off"
        print("light is", self.status)

l = light()
l.turn_on()
l.turn_off()
  
```

Q3 Create a class ShoppingCart:
constructor → total = 0
method add_item(price) → adds to total
method checkout() → prints final bill

```python
class shoppingcart:
    def __init__(self):
        self.total = 0

    def add_item(self,price):
        self.total += price

    def checkout(self):
        print("Total bill is", self.total)

cart = shoppingcart()
cart.add_item(500)
cart.add_item(300)
cart.checkout()
```

Q4 Create a class Employee:
constructor → name & salary
method increase_salary(amount)
method display()

```python
class employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def increase_salary(self, amount):
        self.salary += amount

    def display(self):
        print(self.name, "earns", self.salary)

e1 = employee("tanveer", 30000)
e1.increase_salary(5000)
e1.display()
    
```
