Q1 Create a class Person with:
variable name
a method greet() that prints: "Hello, my name is ___"

```python
class person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print("Hello, my name is", self.name)

p1 = person("Tanveer")
p1.greet()
```

Q2 Create a class Student with:
constructor (__init__) → name & marks
method display() → prints name and marks

```python
class student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    
    def display (self):
        print(self.name, self.marks)

s1 = student("Tanveer", 90)
s1.display()
```

Q3 Create a class BankAccount with:
constructor → balance
method deposit(amount) → adds money
method show_balance() → prints balance

```python
class bankaccount:
    def __init__(self,balance):
        self.balance = balance

    def deposit(self,amount):
        self.balance += amount

    def show_balance(self):
        print("balance is", self.balance)
acc = bankaccount(1000)
acc.deposit(500)
acc.show_balance()
```

Q4 Create a class Rectangle with:
constructor → length & width
method area() → returns area
print the area of rectangle


```python
class rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width
    
r1 = rectangle(10,5)
print("area is ", r1.area())
```
