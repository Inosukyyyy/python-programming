Q3 Create a class Laptop with:
brand and price
use a constructor (__init__) to set values
create 2 objects and print their details

```python
class laptop:
    def __init__(self,brand,price):
        self.brand = brand
        self.price = price

l1 = laptop("HP", 50000)
l2 = laptop("dell",60000)

print(l1.brand,l1.price)
print(l2.brand,l2.price)
```

Q4 Create a class Mobile with:
a constructor that takes brand and price
create 2 objects
print this output format:

```python
class mobile:
    def __init__(self,brand,price):
        self.brand = brand
        self.price = price

m1 = mobile("apple",80000)
m2 = mobile("samsun",50000)

print(m1.brand,m1.price)
print(m2.brand,m2.price)

```
