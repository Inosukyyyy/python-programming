Q2 Basic Constructor
Modify the Student class:
add a constructor (__init__)
initialize name and age through it
create 2 students and print their data

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Student("Tanveer", 20)
s2 = Student("Ashish", 21)

print(s1.name, s1.age)
print(s2.name, s2.age)
```

Q3 Create a class Car:
attributes: brand, speed
method: display_info() that prints both values
create an object and call the method

```python
class car:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

    def display_info(self):
        print("brand:", self.brand)
        print("speed:", self.speed)
        

c1 = car("bmw", 180)
c1.display_info()
```

Q4 Create a class Rectangle:
attributes: length, width
method to calculate area
method to calculate perimeter
create object and print both

```python
class rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

r1 = rectangle(10, 5)

print("Area", r1.area())
print("perimeter", r1.perimeter())
```
