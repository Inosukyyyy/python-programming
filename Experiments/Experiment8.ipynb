### 1. Add few names, one name in each row, in “name.txt file”.
a. Count no of names

b. Count all names starting with vowel

c. Find longest name

```python
with open("name.txt", "w")as f:
    f.write("Tanveer\n")
    f.write("Bhavya\n")
    f.write("Aryan\n")
    f.write("Ashish\n")
    f.write("Ritika\n")
    f.write("Vaanya\n")

with open("name.txt", "r") as f:
    names = f.read().splitlines()

#a)
count_names = len(names)
print("total number of names", count_names)

#b)
vowels = "AEIOUaeiou"
vowel_count = 0 

for name in names:
    if name[0] in vowels:
        vowel_count += 1
print("names starting with vowel:", vowel_count)

#c)
longest_name = max(names, key=len)
print("longest name:", longest_name)
```

### 2. Store integers in a file.
a. Find the max number

b. Find average of all numbers

c. Count number of numbers greater than 100

```python
with open("int.txt", "w") as f:
    f.write("45\n")
    f.write("30\n")
    f.write("150\n")
    f.write("90\n")
    f.write("600\n")
    f.write("12\n")

with open("int.txt", "r") as f:
    numbers = [int(num) for num in f.read().splitlines()]

#a)
max_number = max(numbers)
print("maximum numbers:", max_number)

#b)
average = sum(numbers)/len(numbers)
print("average", average)

#c)
count = 0 
for num in numbers:
    if num > 100:
        count += 1
print("numbers greater than 100:", count)
```

### 4. Input two values from user where the first line contains N, the number of testcases. The next N lines contain the space separated values of a and b. Performinteger division and print a/b. Handle exception in case of ZeroDivisionError orValueError.
Sample input

1 0

2 $

3 1

Sample Output:

Error Code: integer division or modulo by zero

Error Code: invalid literal for int() with base 10: '$' 

```python
n = int(input())

for _ in range(n):
    try: 
        a,b = input().split()
        result = int(a) // int(b)
        print(result)

    except ZeroDivisionError as e:
        print("error", e)
    except ValueError as e:
        print("error", e)

```

### 5. Create multiple suitable exceptions for a file handling program.

```python
try:
    filename=input("Enter filename:")
    with open(filename, "r") as f:
        content=f.read()
        if not content:
            raise ValueError("file is empty")
        
except FileNotFoundError as e:
    print("Error:", e)
except PermissionError as e:
    print("Error:", e)
except Exception:
    print("unexpected error occured")
```
