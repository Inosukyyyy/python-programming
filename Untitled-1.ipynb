# Variable – Native Python?

### A variable is a named memory location that stores data. Its value can change during program execution.

Rules & Guidelines for Variable Names
***
- Can contain letters, digits, and underscores
- Must not start with a digit (✔ name1 / ✖ 1name)
- No spaces allowed (use underscores instead)
- Don’t use Python keywords or built-ins (e.g., print, class, if)
- Names should be short and descriptive (✔ student_name / ✖ s_n)
- Avoid using l (lowercase L) and O (uppercase O) since they look like 1 and 0

# <span style="color: rgb(221, 17, 82);">Python String Built-in Methods</span>



## 1) <span style="color: rgb(221, 17, 82);">Case Conversion Methods</span>



`lower()`
### Definition and Usage
Returns a copy of the string with all alphabetic characters converted to lowercase.
 

```python
"TANVEER".lower()
```

`upper()`
 ### Definition and Usage
 Returns a new string where every cased character has been converted to uppercase.
Keep in mind that s.upper().isupper() may still return False if the string contains uncased characters, or if the uppercase result falls under a Unicode category other than “Lu” (such as “Lt” for titlecase letters).


```python
"python".upper()
```

`title()`
# Definition and Usage
 Returns a version of the string where each word begins with an uppercase character and the remaining characters are lowercase.

```python
"python programming".title()
```

`capitalized()`
# Definition and Usage
Returns a copy of the string with the first character converted to titlecase and the rest converted to lowercase.

Changed in Python 3.8:
The first character is now titlecased instead of strictly uppercased.
This affects digraphs and certain Unicode characters, where only the first letter of the grapheme cluster is capitalized.

```python
"tanveer".capitalize()
```

`swapcase()`
# Definition and Usage
Returns a copy of the string with uppercase characters converted to lowercase and vice versa.


```python
"PyThOn".swapcase()
```

`Casefold()`
 # Definition and Usage
 Returns a caseless, aggressive lowercase version of the string designed for case-insensitive comparisons.
It applies more Unicode transformations than lower(), making it better for text matching across languages.

```python
"Straße".casefold()
```

## <span style="color: rgb(221, 17, 82);">2) Remove Spaces / Cleanup</span>

`strip(chars=None)`

### Definition and Usage

Creates a new string with leading and trailing characters removed.
The chars argument defines which characters may be discarded.
If chars is omitted or set to None, the method trims whitespace by default.
Remember: chars represents a set of characters, not a literal substring.


```python
"  hello  Tanveer    ".strip()
```

`lstrip(chars=none)`

# Definition and Usage

Produces a new string with characters removed from the start (left side).
The chars argument specifies which characters are eligible for removal.
If chars is omitted or set to None, leading whitespace is removed by default.
Note that chars is treated as a set of characters, not a prefix, so any of its characters may be stripped in any order.

```python
"  hello".lstrip()
```

`rstrip(chars=None)`

# Definition and Usage

Returns a new string with characters trimmed from the end (right side).
The chars parameter defines which characters can be removed.
If chars is not provided or set to None, trailing whitespace is removed by default.
Remember that chars acts as a set of characters rather than a suffix, so any combination of those characters may be stripped.

```python
"hello   ".rstrip()   # 'hello'
```
