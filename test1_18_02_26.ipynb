### Q2

```python
def count_by_category(pairs):
    d = {}
    for cat, word in pairs:
        d.setdefault(cat, {})
        d{cat}.setdefault(word, 0)
        d[cat][word] += 1
    return d
```

### Q3

```python
def missing_number(nums):
    n = len(nums)
    total = n * (n+1) // 2
    return total - sum(nums)
```

```python
def add_unique(x,seen=None):
    if seen is None:
        seen = set()
    if x not in seen:
        seen.add(x)
    return list(seen)
    
```
